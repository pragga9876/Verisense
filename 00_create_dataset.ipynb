```python
import pandas as pd

# Load the two CSV files
fake_df = pd.read_csv('Fake.csv')
true_df = pd.read_csv('True.csv')

# Add labels
fake_df['label'] = 'fake'
true_df['label'] = 'real'

# Combine them
combined_df = pd.concat([fake_df, true_df], ignore_index=True)

# Shuffle the data (so real and fake are mixed)
combined_df = combined_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Keep only what we need
final_df = combined_df[['text', 'label']]

# Save as news.csv
final_df.to_csv('news.csv', index=False)

# Verify it worked
print(f"✅ SUCCESS! Created news.csv with {len(final_df)} articles")
print(f"   Real articles: {len(final_df[final_df['label'] == 'real'])}")
print(f"   Fake articles: {len(final_df[final_df['label'] == 'fake'])}")
print("\n📝 First article preview:")
print(final_df['text'].iloc[0][:200])
print("...")
```
