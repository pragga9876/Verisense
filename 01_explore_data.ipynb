```python
import pandas as pd

df = pd.read_csv('news.csv')

print("=" * 50)
print("DATA EXPLORATION")
print("=" * 50)

print(f"\n📊 Total articles: {len(df)}")
print(f"✅ Real: {len(df[df['label'] == 'real'])}")
print(f"❌ Fake: {len(df[df['label'] == 'fake'])}")

print("\n" + "=" * 50)
print("SAMPLE REAL NEWS:")
print("=" * 50)
print(df[df['label'] == 'real']['text'].iloc[0])

print("\n" + "=" * 50)
print("SAMPLE FAKE NEWS:")
print("=" * 50)
print(df[df['label'] == 'fake']['text'].iloc[0])

print("\n" + "=" * 50)
print("COUNT EXCLAMATION MARKS (Fake news uses more !!!)")
print("=" * 50)

real_text = df[df['label'] == 'real']['text'].iloc[0]
fake_text = df[df['label'] == 'fake']['text'].iloc[0]

print(f"Real news exclamation marks: {real_text.count('!')}")
print(f"Fake news exclamation marks: {fake_text.count('!')}")
```
